```python
import math

n1, n2, n3 = 6, 11, 35
d1, d2, d3 = 5, 3, 4

def modulo_inverse(A, B):
    """
        Find the inverse of A while modulo B
        E.g 
            - inverse of 31 is 26 while modulo 35
            - inverse of 7 is 5 while modulo 17
    """
    res = 1
    for inv in range(1, B + 1):
        if inv * A % B == 1:
            res = inv
    return res

def solve(n1, n2, n3, d1, d2, d3):
    """
        PROBLEM STATEMENT : 
            Find x such that :
                x mod n1 = d1
                x mod n2 = d2
                x mod n3 = d3            
    """
    N = [n1, n2, n3]
    D = [d1, d2, d3]
    S = 0

    if math.gcd(n1, n2, n3) == 1:    
        M = math.lcm(n1, n2, n3)
        for idx in range(3):
            temp = M // N[idx]
            r = temp % N[idx]
            r_inv = modulo_inverse(r, N[idx])
            print(f"M[{idx + 1}] = {temp} \t r[{idx + 1}] = {r} \t (r[{idx + 1}])^(-1) = {r_inv}")

            S += D[idx] * temp * r_inv

        return S % M

solve(n1, n2, n3, d1, d2, d3)
```
